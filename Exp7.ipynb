```python
import pandas as pd

# Load dataset
url = 'https://gist.githubusercontent.com/curran/a08a1080b88344b0c8a7/raw/0e7a9b0a5d22642a06d3d5b9bcbad9890c8ee534/iris.csv'
df = pd.read_csv(url)
```

```python
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
print(df.info())
```

```python
# Separate numerical and categorical columns
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

print("Numerical Columns:", numerical_cols)
print("Categorical Columns:", categorical_cols)
```

```python
from sklearn.preprocessing import StandardScaler

# Standardize numerical features
scaler = StandardScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

# Display the dataset with scaled features
print(df.head())
```

```python
# One-Hot Encoding for categorical columns
df_encoded = pd.get_dummies(df, columns=categorical_cols)

# Display dataset with encoded features
print(df_encoded.head())
```

```python
# Compare original and preprocessed dataset
print("Original Dataset:")
print(df.head())

print("Preprocessed Dataset:")
print(df_encoded.head())
```

```python
print(f"Memory usage before preprocessing: {df.memory_usage(deep=True).sum()} bytes")
print(f"Memory usage after preprocessing: {df_encoded.memory_usage(deep=True).sum()} bytes")
```
